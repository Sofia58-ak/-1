### Задание 7. Гистограмма

**Ответы на контрольные вопросы:**
1. **Что такое DataFrame?**  
   Это двумерная табличная структура данных в библиотеке Pandas, состоящая из строк и столбцов, где каждый столбец может иметь свой тип данных.
2. **Что такое признак (Feature)?**  
   Это отдельный столбец в датасете, который описывает определенную характеристику или свойство объекта наблюдения (например, `popularity`, `genre`, `release_year`).
3. **Чем среднее арифметическое отличается от медианы?**  
   Среднее арифметическое чувствительно к экстремальным значениям (выбросам), так как учитывает все числа при сложении. Медиана показывает центральное значение в упорядоченном ряду и не искажается выбросами.
4. **Что показывает гистограмма?**  
   Гистограмма показывает распределение частот данных, позволяя визуально оценить форму распределения, центр, разброс и наличие аномальных значений (выбросов).
5. **Для чего нужна фильтрация данных?**  
   Фильтрация нужна для отбора подмножества данных, удовлетворяющих определенным условиям, чтобы анализировать только релевантные для конкретной задачи записи (например, только новые треки или только определенный жанр).
